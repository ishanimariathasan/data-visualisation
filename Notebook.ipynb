```python
if (!requireNamespace("tidyverse", quietly = TRUE)) install.packages("tidyverse")
if (!requireNamespace("DT", quietly = TRUE)) install.packages("DT")
library(tidyverse)
library(knitr)
library(DT)

dir.create("data/cleaned", recursive = TRUE, showWarnings = FALSE)
dir.create("figures", recursive = TRUE, showWarnings = FALSE)
```

```python
# Load Age_onMale data using same approach as BarPlot.r
df <- read.csv("Age_onMale.csv", header = FALSE, skip = 2, stringsAsFactors = FALSE)
names(df) <- c("label", "sample", "young", "old")
df <- df[, c("sample", "young", "old")]
df$young <- as.numeric(df$young)
df$old <- as.numeric(df$old)

glimpse(df)
dim(df)
sum(is.na(df))
```

```python
# Full inline table (interactive)
DT::datatable(df, options = list(pageLength = 100, scrollX = TRUE))
```

```python
# Minimal cleaning (trim whitespace, convert empty strings to NA)
df_clean <- df %>%
	mutate(across(where(is.character), ~ na_if(trimws(.), ""))) %>%
	mutate(across(c(young, old), as.numeric))

readr::write_csv(df_clean, "data/cleaned/Age_onMale_cleaned.csv")

cat("Wrote cleaned CSV to data/cleaned/Age_onMale_cleaned.csv\n")
```

```python
# Source BarPlot.r (will create and save a figure to figures/)
source("BarPlot.r")

# Display saved image inline if present
if (file.exists("figures/age_on_male_barplot.png")) {
	knitr::include_graphics("figures/age_on_male_barplot.png")
} else {
	cat("Plot image not found in figures/. Check BarPlot.r execution.\n")
}
```

```python
# List generated outputs
list.files("data/cleaned", full.names = TRUE)
list.files("figures", full.names = TRUE)

cat("Done. Cleaned data and figures are in data/cleaned/ and figures/.\n")
```
